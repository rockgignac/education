## Data Preparation & Cleaning

```python
school_information = school_information.rename(
    columns={
    "SCHOOL_YEAR": "year", 
    "NCESSCH": "id", 
    "LSTATE": "state", 
    "LZIP": "zip_code", 
    "SCH_TYPE_TEXT": "school_type", 
    "LEVEL": "school_level",
    "CHARTER_TEXT": "charter"
})
```

```python
school_information = school_information[
    ['year', 'id', 'state', 'zip_code', 'school_type', 'school_level','charter']
]
```

```python
school_information.head()
```

```python
edgap = edgap.rename(
    columns={
        "NCESSCH School ID": "id",
        "CT Pct Adults with College Degree": "percent_college",
        "CT Unemployment Rate": "rate_unemployment",
        "CT Pct Childre In Married Couple Family": "percent_married",
        "CT Median Household Income": "median_income",
        "School ACT average (or equivalent if SAT score)": "average_act",
        "School Pct Free and Reduced Lunch": "percent_lunch",
    }
)
```

```python
edgap.head()
```

```python
school_information = school_information.rename(columns={"NCESSCH": "id"})
```

```python
school_information.info()
```

```python
school_information = school_information.dropna(subset=['id'])
school_information['id'] = school_information['id'].astype('int64')
edgap['id'] = edgap['id'].astype('int64')
```

```python
df = edgap.merge(
    school_information,
    how = 'left',
    on = "id"
)
```

```python
df['id'] = df['id'].astype(str)
```

```python
df.head()
```

```python
df.info()
```

## Quality Control

```python
df.describe()
```

```python
df.select_dtypes(include=['number']).agg(['min', 'max']).round(2)
```

```python
df.loc[df['percent_lunch'] < 0, 'percent_lunch'] = np.nan
```

```python
df.loc[df['average_act'] < 1, 'average_act'] = np.nan
```

```python
df = df.loc[df['school_level'] == 'High']
```

```python
df.duplicated().sum()
```

```python
df.isna().sum().to_frame(name='Number of Missing Values')
```

```python
import plotly.offline as po
import plotly.graph_objs as pg
```

```python
layout = dict(
    geo={"scope": "usa"}, coloraxis_colorbar=dict(title="Number of Schools")
)

data = dict(
    type="choropleth",
    locations=df["state"].value_counts().index,
    locationmode="USA-states",
    z=df["state"].value_counts().values,
    coloraxis="coloraxis",
)

x = pg.Figure(data=[data], layout=layout)

po.iplot(x)
```

```python
df = df.dropna(subset=['average_act'])
```

```python
df.isna().sum().to_frame(name='Number of Missing Values')
```

```python
predictor_variables = [
    'rate_unemployment',
    'percent_college',
    'percent_married',
    'median_income',
    'percent_lunch',
    'state',
    'charter'
]
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
imputer = IterativeImputer()
```

```python
numerical_predictors = df[predictor_variables].select_dtypes(include='number').columns.to_list()

print(numerical_predictors)
```

```python
imputer.fit(df.loc[:, numerical_predictors])
```

```python
df.loc[:, numerical_predictors] = imputer.transform(df.loc[:, numerical_predictors])
```

```python
df.isna().sum().to_frame(name='Number of Missing Values')
```

```python
df.to_csv(
    'education_clean.csv',
    encoding='utf-8-sig',
    index=False
)
```
